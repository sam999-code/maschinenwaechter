# استكشاف إشارة الاهتزاز

هذا الدفتر هو نقطة البداية لاستكشاف بيانات أجهزة الاستشعار المحاكاة من المشروع **maschinenwaechter**.

سنحمل عينة CSV، ننظر إلى الإحصاءات الوصفية، ثم نرسم إشارة الاهتزاز مع تحديد منطقة العطل.

*Vibration signal exploration — load a simulated machine CSV, run `describe()`, and plot the vibration channel.*

```python
import sys
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt

project_root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
df = pd.read_csv(project_root / "data" / "sample" / "machine_01.csv", parse_dates=["timestamp"])
df.describe()
```

## رسم إشارة الاهتزاز

نرسم إشارة الاهتزاز (بالملم/ثانية) عبر الزمن، ونظلّل المنطقة التي بدأ فيها العطل باللون الأحمر.

*Plot vibration (mm/s) over time with the fault region shaded red.*

```python
fig, ax = plt.subplots(figsize=(14, 4))
ax.plot(df["timestamp"], df["vibration_mm_s"], linewidth=0.4)
fault_start = df.loc[df["fault"] == 1, "timestamp"].min()
ax.axvspan(fault_start, df["timestamp"].max(), color="red", alpha=0.10)
ax.set_ylabel("vibration (mm/s)")
ax.set_title("Vibration signal — fault region shaded")
plt.show()
```
